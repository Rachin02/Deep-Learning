```python
import tensorflow 
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Embedding, Dense, SimpleRNN, LSTM, Bidirectional
from tensorflow.keras.models import Sequential
from tensorflow.keras.datasets import imdb
```

```python
(x_train, y_train),((x_test, y_test)) = imdb.load_data(num_words = 6000)
```

```python
x_train = pad_sequences(x_train, maxlen = 100)
x_test = pad_sequences(x_test, maxlen = 100)
```

```python
model = Sequential()
model.add(Embedding(6000,32,input_shape = (100,)))
model.add(SimpleRNN(5))
model.add(Dense(1, activation = 'sigmoid'))

model.summary()
```

```python
model.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics =['accuracy'])
```

```python
model.fit(x_train,y_train, epochs = 7, validation_data=(x_test,y_test))
```

```python
model = Sequential()
model.add(Embedding(6000,32,input_shape = (100,)))
model.add(Bidirectional(SimpleRNN(5)))
model.add(Dense(1, activation = 'sigmoid'))

model.summary()
```

```python
model.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics =['accuracy'])
```

```python
model.fit(x_train,y_train, epochs = 7, validation_data=(x_test,y_test))
```
