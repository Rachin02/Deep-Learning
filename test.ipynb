## Linklist

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
a = Node(10)
b = Node(20)
c = Node(30)
a.next= b
b.next= c
```

```python
current = a
while current:
    print(current.data)
    current = current.next
```

## Binary Tree

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
```

```python
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
```

```python
print(root.data)
print(root.right.data)
print(root.left.right.data)
```

## DFS

```python
def inorder(root):

    if root:
        inorder(root.left)
        print(root.data)
        inorder(root.right)
```

## BFS

```python
from collections import deque

def bfs(root):

    queue=deque()
    queue.append(root)


    while queue:
        node=queue.popleft()
        print(node.data)

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)
```

```python
## Linear Search = O(n)
def linear_search(arr,target):

    for i in range(len(arr)):

        if arr[i] == target:
            return i

    return -1



arr=[5,8,3,9,2]

print(linear_search(arr,9))
```

```python
## Bubble sort

li = [5,3,9,6,1,8,2]
n = len(li)

for i in range(0, n):
    for j in range(0 , n-1):
        if li[j] > li[j+1]:
            temp = li[j]
            li[j] = li[j+1]
            li[j+1] = temp

print(li)
```

```python
## Selection sort

li = [5,3,9,6,1,8,2]
n = len(li)




for i in range(0, n):
    min_value = i
    for j in range(i, n):
        if li[j] < li[min_value]:
            min_value = j

    temp = li[i]
    li[i] = li[min_value]
    li[min_value] = temp

print(li)
    
```

```python
## Insertion Sort

li = [5,3,9,6,1,8,2]
n = len(li)

for i in range(1, n):
    val = li[i]
    j = i -1

    while j >= 0 and li[j] > val:
        li[j+1] = li[j]
        j = j-1

    li[j+1] = val

print(li)
```
