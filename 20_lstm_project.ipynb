```python
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([data])
```

```python
len(tokenizer.word_index)
```

```python
data_to_sequence = []

for sentence in data.split('\n'):
    token = tokenizer.texts_to_sequences([sentence])[0]

    for i in range(1, len(token)):
        data_to_sequence.append(token[:i+1])
```

```python
data_to_sequence
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
max_len = max(len(line) for line in data_to_sequence)

input_sequence = pad_sequences(data_to_sequence, maxlen = max_len, padding = 'pre')
```

```python
x = input_sequence[:,:-1]
y = input_sequence[:,-1]
```

```python
x.shape, y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y, num_classes = 521)
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
x_train.shape,  y_train.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, LSTM,Dropout
```

```python
model = Sequential()
model.add(Embedding(521, 128, input_shape = (max_len-1,)))
model.add(LSTM(128, return_sequences= True))
model.add(Dropout(0.2))
model.add(LSTM(84))
model.add(Dropout(0.2))
model.add(Dense(521, activation = 'softmax'))

model.summary()
```

```python
model.compile(loss = 'categorical_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
```

```python
history = model.fit(x_train, y_train, epochs = 90, validation_data=(x_test,y_test))
```

```python
import time

word = 'Machine'

for i in range(10):
    # tokens
    tokens = tokenizer.texts_to_sequences([word])[0]
    # padding
    input_padding = pad_sequences([tokens], maxlen = max_len, padding = 'pre')

    #predict
    y_pred = model.predict((input_padding))
    y_pred = y_pred.argmax(axis = 1)

    for text, index in tokenizer.word_index.items():
        if y_pred == index:
            word = word+' '+text
            print(word)

            time.sleep(1)
    
```
